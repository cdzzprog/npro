```python
import os
import geopandas as gpd
import pandas as pd

# 定义输入文件夹路径和输出文件夹路径
input_folder_1 = r'C:\Users\龙儿璨\Desktop\湿地制图\3'
input_folder_2 = r'C:\Users\龙儿璨\Desktop\湿地制图\4'
output_folder = r'C:\Users\龙儿璨\Desktop\湿地制图\5'

# 确保输出文件夹存在
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# 遍历第一个文件夹中的所有shapefile文件
for file_name in os.listdir(input_folder_1):
    if file_name.endswith('.shp'):
        # 构建两个输入文件夹中对应文件的完整路径
        file_1 = os.path.join(input_folder_1, file_name)
        file_2 = os.path.join(input_folder_2, file_name)  # 假设两个文件夹中有对应文件名
        
        # 确保第二个文件夹中存在对应的矢量文件
        if os.path.exists(file_2):
            # 读取第一个文件夹中的矢量文件
            gdf_1 = gpd.read_file(file_1)
            # 读取第二个文件夹中的矢量文件
            gdf_2 = gpd.read_file(file_2)
            
            # 合并第一个文件中的所有要素（如果需要单个文件的要素合并）
            merged_gdf_1 = gdf_1.unary_union
            gdf_1_merged = gpd.GeoDataFrame(geometry=[merged_gdf_1], crs=gdf_1.crs)
            
            # 合并第二个文件中的所有要素
            merged_gdf_2 = gdf_2.unary_union
            gdf_2_merged = gpd.GeoDataFrame(geometry=[merged_gdf_2], crs=gdf_2.crs)
            
            # 将两个合并后的矢量文件进行合并
            combined_gdf = gpd.GeoDataFrame(pd.concat([gdf_1_merged, gdf_2_merged], ignore_index=True), crs=gdf_1.crs)
            
            # 添加一个名为 'label' 的空字段
            combined_gdf['label'] = None  # 创建一个空的label列
            
            # 构建输出文件的路径
            output_file = os.path.join(output_folder, file_name)
            
            # 将合并后的矢量文件保存为新的Shapefile
            combined_gdf.to_file(output_file)
            
            print(f"Combined shapefile saved with label column: {output_file}")
```
